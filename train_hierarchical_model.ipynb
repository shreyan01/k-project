# Hierarchical Prediction & Guidance Model

## New Approach: Protocol-Aware Prediction with Error Detection

This notebook implements a new architecture that addresses FineBio's core challenges:

### Key Innovations:
1. **Protocol-Aware Transformer**: Respects strict step ordering but allows flexible atomic operations
2. **Graph Neural Network**: Models spatial relationships between objects (hand-object, object-object)
3. **Multi-Task Learning**: Simultaneously predicts steps, atomic ops, and next steps
4. **Error Detection**: Contrastive learning to detect when user deviates from expected sequence
5. **Guidance Generation**: Provides actionable feedback when errors are detected

### Why This Approach Works:
- **Handles variable atomic operation ordering**: Attention mechanism learns flexible patterns
- **Better hand-object understanding**: Graph features capture spatial relationships
- **Real-time guidance**: Error detection triggers immediate feedback
- **Future prediction**: Next-step prediction enables proactive guidance
